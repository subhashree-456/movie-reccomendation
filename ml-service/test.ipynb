```python
#keep important columns 
movies = movies[['id' , 'title' , 'vote_average' , 'overview' , 'tagline' , 'genres' , 'spoken_languages' ]]
```

```python
movies = movies.dropna(subset=['title'])
```

```python
movies['overview'] = movies['overview'].fillna('')
```

```python
movies['tagline'] = movies['tagline'].fillna('')
```

```python
movies['genres'] = movies['genres'].fillna('')
```

```python
movies['genres'] = movies['genres'].apply(lambda x: x.replace(",", ""))
```

```python
movies['spoken_languages'] = movies['spoken_languages'].fillna('')
movies['spoken_languages'] = movies['spoken_languages'].apply(lambda x: x.replace(",", ""))
```

```python
movies['tags'] = movies['overview'] + " " + movies['tagline'] + " " + movies['genres'] + " " + movies['spoken_languages']
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import re
```

```python
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()
```

```python
def preprocess_text(text) : 
    text = str(text).lower()
    #remove the punctuations 
    text = re.sub(r'[^a-zA-Z\s]' , '' , text)

    words = text.split()
    #remove stop words 
    words = [ word for word in words if word not in stop_words]
    #lemmatize 
    words = [lemmatizer.lemmatize(word) for word in words]

    return " ".join(words)
```

```python
movies['tags'] = movies['tags'].apply(preprocess_text)
```

```python
movies['tags'][0]
```

```python
movies = movies.reset_index(drop=True)
```

```python
indices = pd.Series(movies.index , index= movies['title']).drop_duplicates()
indices
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf = TfidfVectorizer(max_features=50000 , ngram_range=(1,2), stop_words='english')
```

```python
tfidf_matrix = tfidf.fit_transform(movies['tags'])
tfidf_matrix
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
def movie_recommendation(title, n=10):
    if title not in indices:
        return ['movie not found!']
    
    idx = indices[title]
    
    sim_score = cosine_similarity(tfidf_matrix[idx], tfidf_matrix).flatten()
    
    # ⚡ efficient top-n (no full sort)
    top_idx = np.argpartition(sim_score, -n-1)[-n-1:]
    top_idx = top_idx[np.argsort(sim_score[top_idx])][::-1]
    
    # remove itself
    top_idx = top_idx[1:n+1]
    
    return movies.iloc[top_idx]['title']
```
